### Exercise 1.2
In the titanic dataset convert columns with qualitative variables
to factors (sex, embarked, survived) or ordered factor (pclass) -
defining the appropriate levels and, if needed, order of values.
Please code the levels as:
- embarked respectively: Cherbourg, Queenstown, Southampton
- sex - Female, Male
- survived - Yes, No
- pclass - 1st <2nd <3rd

```python
embar = {'C': 'Cherbourg', 'Q': 'Queenstown', 'S': 'Southampton'}
s = {0:'Male', 1:'Female'}
pc = {1:'1st', 2:'2nd', 3:'3rd'}
```

```python
titanic['embarked'] = titanic.embarked.map(embar)
#titanic['sex'] = titanic.sex.map(s)
#titanic['pclass'] = titanic.pclass.map(pc)
print(titanic['embarked'].unique())
```

```python
titanic['pclass'] = titanic.pclass.map(pc)
print(titanic['pclass'].unique())
```

```python
titanic['sex'] = titanic.sex.map(s)
print(titanic['sex'].unique())
```

```python
titanic['survived'] = titanic.survived.map({0: 'No', 1:'Yes'})
print(titanic['survived'].unique())
```
